```python
import numpy as np
import matplotlib.pyplot as plt
# Generate data using NumPy arrays
x = np.array([0, 1, 2, 3, 4, 5])
y = x ** 2
# Create a plot using Matplotlib with specified marker, marker size, line style, and line width
plt.plot(x, y, marker='o', markersize=8, linestyle=':',linewidth=2, color='#FF5733', label='Data Points')
# Customize plot
plt.title('Plot with Customizations')
plt.xlabel('x values')
plt.ylabel('y values')
plt.grid(True)
# Customize x and y ticks
plt.xticks(np.arange(0, 6, step=1)) # Custom x ticks from 0 to 5 with step size 1
plt.yticks(np.arange(0, 26, step=5)) # Custom y ticks from 0 to 25 with step size 5
# Display legend
plt.legend()
# Display the plot
plt.show()
```

```python
categories = ['A', 'B', 'C', 'D']
values = [10, 15, 7, 12]
plt.bar(categories,values, color = 'blue', edgecolor = 'red')
plt.xlabel('categories')
plt.ylabel('values')
plt.title('Basic Example of a bar char')
plt.show()
```

```python
categories = ['A', 'B', 'C', 'D']
values = [10, 15, 7, 12]
plt.barh(categories,values, color = 'blue', edgecolor = 'red')
plt.xlabel('categories')
plt.ylabel('values')
plt.title('Basic Example of a bar char')
plt.show()
```

```python
values1 = [10, 15, 7, 12]
values2 = [8, 12, 5, 10]
plt.bar(categories, values1, color='blue', label='Furniture')
plt.bar(categories, values2, bottom = values1, color='orange', label='Technology')
plt.legend()
plt.show()
```

```python
cars = ['AUDI', 'BMW', 'FORD',
        'TESLA', 'JAGUAR', 'MERCEDES']
data = [23, 17, 35, 29, 12, 41]
plt.pie(data, labels = cars)
plt.show()
```

```python
cars = ['AUDI', 'BMW', 'FORD',
        'TESLA']
data = [23, 17, 35, 29]
e = (1,0.2,0,0)
plt.pie(data, labels = cars, explode=e)
plt.title("Pie chart")
plt.show()
```

```python
labels = ['A', 'B', 'C', 'D']
values = [30, 20, 45, 75]
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']
plt.pie(values, labels=labels, colors = colors, autopct='%1.1f%%',startangle=90, wedgeprops={'linewidth': 1, 'edgecolor': 'white'})

center_cicle = plt.Circle((0,0),0.70,color ='white')
plt.gca().add_artist(center_cicle)
plt.show()
```
