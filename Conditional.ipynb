```python
#Nested while Loop:
outer_count = 0
while outer_count < 3:
    print("Outer loop iteration:", outer_count)
    inner_count = 0
    while inner_count < 2:
        print("Inner loop iteration:", inner_count)
        inner_count += 1
    outer_count += 1
```
