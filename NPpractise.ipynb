```python
import numpy as np
```

```python
#1st array
my_array = np.array([1,2,3,4])
my_array
```

```python
print(type(my_array))
```

```python
my_array.dtype
```

```python
#2d array
My2nd = np.array([[1,2,3],[5,6,7]])
My2nd
```

```python
#3d array
My3d = np.array([[[1,2,3,4],[5,6,7,8]],[[8,9,10,12],[14,16,17,19]]])
My3d
```

```python
My2 = np.array([[1,2,3],[5,6,7]],dtype=float)
My2
```

```python
#Dimensin
print(my_array.ndim)
print(My2nd.ndim)
print(My3d.ndim)
```

```python
#Shape
print(my_array.shape)
print(My2nd.shape)
print(My3d.shape)
```

```python
#Size
print(my_array.size)
print(My2nd.size)
print(My3d.size)
```

```python
#datatype
print(my_array.dtype)
print(My2nd.dtype)
print(My2.dtype)
```

```python
#itemsize
print(my_array.itemsize)
print(My2nd.itemsize)
print(My3d.itemsize)
```

```python
#memory location
print(my_array.data)
print(My2nd.data)
print(My3d.data)
```

```python
#Index and Slicing
#Index
print(my_array)
print(my_array[2])
print(My2nd)
print(My2nd[1,2])
print(My3d)
print(My3d[1,0,2])
```

```python
#Bool index
print(my_array)
bool_value = [True,False,True,True]
print(my_array[bool_value])
```

```python
#Slicing
#1D
print(my_array[1:3])
print(my_array[::-1])
print(my_array[::2])
print()
#2D
New2d = np.array([[1,2,3,4],[6,7,8,9],[3,8,9,10],[2,3,4,5]])
print(New2d)
print(New2d[1:3,1:3])
print(New2d[:2])
print(New2d[:,1:])
print(New2d[:,:-1])
```

```python
#3D
print(My3d)
print(My3d[0,1,1:3])
print(My3d[...,:])
print(My3d[:,0,...])
```

```python
#Operations:
First = np.array([1,2,3,4])
Sec = np.array([5,6,7,8])
print(First)
print(Sec)
```

```python
D = First+Sec
print(D)
E = First-Sec
print(E)
F = First*Sec
print(F)
G = First/Sec
print(G)
H = First//Sec
print(H)
I = First%Sec
print(I)
L = First**Sec
print(L)
```

```python
#Transpose
print(New2d)
print(New2d.T)
print(np.sort(New2d))
print(np.sort(New2d,axis=0))
print(np.sort(New2d,axis=1))
print(np.sort(New2d,axis=-1))
```

```python
#conact
result = np.concatenate((First,Sec))
print(result)
```

```python
fi2nd = np.array([[1,2,3],[4,5,6]])
sec2nd = np.array([[4,5,6],[7,8,9]])
print(np.concatenate((fi2nd,sec2nd),axis=0))
print(np.concatenate((fi2nd,sec2nd),axis=1))
```

```python
print(np.split(My3d,2,axis=0))
print(np.split(My3d,2,axis=1))
```

```python
#Stat operations
print(New2d)
print(np.mean(New2d,axis=0))
print(np.mean(New2d,axis=1))
print(np.median(New2d,axis=0))
print(np.median(New2d,axis=1))
print(np.std(New2d,axis=0))
print(np.std(New2d,axis=1))
print(np.sum(New2d,axis=0))
print(np.sum(New2d,axis=1))
print(np.max(New2d,axis=0))
print(np.max(New2d,axis=1))
print(np.min(New2d,axis=0))
print(np.min(New2d,axis=1))
```

```python
#Types of Metrix
print(np.zeros(5))
print(np.zeros((3,4)))
print(np.zeros((3,4,2)))
print(np.ones(5))
print(np.ones((3,4)))
print(np.ones((3,4,2)))
print(np.full((3,4),-1))
print(np.eye(3))
print(np.eye(4,k=2))
```

```python
print(np.arange(10))
print(np.arange(3,13))
print(np.arange(0,11,2))
```

```python
print(np.random.random(5))
print(np.random.random((2,3)))
```
