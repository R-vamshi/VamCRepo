```python
#Use of key in lambda sorted
# Purpose of lambda Functions: lambda functions allow for quick, anonymous function
# definitions, useful when we need a simple function temporarily, especially within functions
# like sorted() or map() where only a small transformation is required.
# How key in sorted() Works: The key parameter allows specifying a function (often a lambda)
# that computes a value for sorting. Instead of comparing items directly, sorted() compares the
# results of applying the key to each item.
# Example Scenario: Sorting a list of dictionaries by a specific field (e.g., age), where each
# dictionary represents a person. Using sorted(people, key=lambda x: x['age']) is simpler and
# more readable than writing a for loop to sort by extracting and comparing age.
```

```python
# compare list List comprehensions and for loop
# Difference: A list comprehension is a compact syntax for generating a new list from an
# existing iterable. It combines looping and conditional logic into a single line, making code
# shorter and often faster since it avoids some overhead associated with a for loop.
# When List Comprehensions are Preferred: When the goal is to create a new list by
# transforming or filtering elements from an iterable, list comprehensions provide a concise,
# readable syntax (e.g., [x ** 2 for x in range(10) if x % 2 == 0]).
# When for Loops are Preferred: If the operation involves complex logic, multiple actions
# within each iteration, or updates to multiple variables, a traditional for loop may be clearer
# and easier to debug, as list comprehensions become harder to read and maintain in these
# cases.
```

```python
import pandas as pd
data = {
'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Hank'],
'Age': [25, 30, 35, 40, 28, 45, 50, 33],
'Salary': [50000, 60000, 55000, 65000, 48000, 70000, 75000, 62000],
'Department': ['HR', 'Finance', 'IT', 'IT', 'Finance', 'HR', 'IT', 'Finance']
}
df = pd.DataFrame(data)
print(df)
```

```python
def cat_age(age):
    if age<30:
        return 'Junior'
    elif 30 <= age <= 40:
        return 'Mid'
    elif age>40:
        return 'Senior'

df['age_group'] = df['Age'].apply(cat_age)
print(df[['Employee','Age','age_group']])
```

```python
df['Salary']=df.apply(lambda row:row['Salary']*1.1 if row['Department']=='IT' else row['Salary'],axis=1)
print(df)
```

```python
df['Tax']=df.apply(lambda row:row['Salary']*0.5 if row['Department']=='HR' else row['Salary']*0.8,axis=1)
print(df)
```

```python
for index,row in df.iterrows():
    print(f"Emp:{row['Employee']},Age:{row['Age']},Sal:{row['Salary']}")
```

```python
dept_sal = {}
for index,row in df.iterrows():
    dept = row['Department']
    if dept not in dept_sal:
        dept_sal[dept]=[]
    dept_sal[dept].append(row['Salary'])

for dept,salary in dept_sal.items():
    avg_sal = sum(salary)/len(salary)
    print(f"avg sal for {dept}:{avg_sal}")
```

```python
dept_salaries = {}
for index, row in df.iterrows():
    dept = row['Department']
    if dept not in dept_salaries:
        dept_salaries[dept] = []
    dept_salaries[dept].append(row['Salary'])

for dept, salaries in dept_salaries.items():
    avg_salary = sum(salaries) / len(salaries)
    print(f"Average Salary in {dept}: {avg_salary}")
```

```python
dept_avg = {dept: sum(salaries) / len(salaries) for dept, salaries in
dept_salaries.items()}
for index,row in df.iterrows():
    if row['Salary'] > dept_avg[row['Department']]:
        print(f"{row['Employee']} has a salary above the department average.")
```

```python
import pandas as pd
data = {
'Department': ['HR', 'Finance', 'IT', 'HR', 'Finance', 'IT', 'Finance', 'IT'],
'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Hank'],
'Age': [25, 30, 35, 40, 28, 45, 50, 33],
'Salary': [50000, 60000, 55000, 65000, 48000, 70000, 75000, 62000]
}
df = pd.DataFrame(data)
print(df)
```

```python
result = df.groupby('Department').agg({'Salary':'sum','Age':'min'})
print(result)
```

```python
result = df.groupby('Department').agg({'Salary':'mean','Age':'min'})
print(result)
```

```python
df['Dept_Avg_Salary']=df.groupby('Department')['Salary'].transform('mean')
df_above_age = df[df['Salary']>df['Dept_Avg_Salary']][['Department','Employee','Salary','Dept_Avg_Salary']]
print(df_above_age)
```

```python
df['Dept_Avg_Age']=df.groupby('Department')['Age'].transform('mean')
df['Age_Relative_to_Avg']=df.apply(lambda x:'Above Avg' if x['Age']>x['Dept_Avg_Age'] else 'Below Avg',axis=1)
print(df[['Department','Employee','Age','Dept_Avg_Age','Age_Relative_to_Avg']])
```

```python
df['Max_Dept_Sal']=df.groupby('Department')['Salary'].transform('max')
df['Pct_Diff_from_Max'] = ((df['Max_Dept_Sal']-df['Salary'])/df['Max_Dept_Sal'])
print(df[['Department','Employee','Age','Max_Dept_Sal','Pct_Diff_from_Max']])
```

```python
def sample_generater(limit):
    num = 0
    while num<limit:
        yield num
        num+=1

for x in sample_generater(7):
    print(x)
```

```python
def squre(limit):
    for x in range(1,limit+1):
        yield x**2

for i in squre(7):
    print(i)
```

```python
def even(limit):
    for num in range(1,limit+1):
        if num%2 == 0:
            yield x

for i in even(10):
    print(i)
```

```python
def generate_factorial(limit):
    fact = 1
    for num in range(1,limit+1):
        fact*=num
        yield fact

for x in generate_factorial(5):
    print(x)
```

```python
depart = {}
for index,row in df4.iterrows():
    dept = row['Department']
    if dept not in depart:
        depart[dept]=[]
    depart[dept].append(row['Salary'])

for dept,sal in depart.items():
    avg_sal = sum(sal)/len(sal)
    print(f'Avg sal of dept {dept}:{avg_sal}')
```

```python
dept_avg = {dept:sum(sal)/len(sal) for dept,sal in depart.items()}
for index,row in df4.iterrows():
    if row['Salary']>dept_avg[row['Department']]:
        print(f'{row['Employee']} having salary greater than avg')
```
