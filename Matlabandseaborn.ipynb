```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x = [1,2,3,4,5]
y=[2,3,5,7,11]
```

```python
plt.plot(x,y,label='Prime number')
plt.title('Simple line plot')
plt.xlabel('Xlabel')
plt.ylabel('Ylabel')
plt.legend()
plt.show()
```

```python
height = ['121.9','124.5','129.5','134.6','139.7','147.3','152.4','157.5','162.6']
weight = ['19.7','21.3','23.5','25.9','28.5','32.1','35.7','39.6','43.2']
plt.figure(figsize=(8, 6))
plt.plot(weight, height, color='green', marker='*', markersize=10,
linestyle='--', linewidth=2, label='Average Weight vs Height')
plt.title('avg weight with respect to avg height')
plt.xlabel('Weight in KG')
plt.ylabel('Height in cm')
plt.grid(True)
plt.legend()
plt.show()
```

```python
x = np.array([0,1,2,3,4,5,6])
y = x**2
```

```python
plt.plot(x,y,marker='o',markersize=8,linestyle=':',linewidth=2,color="#FF5733",label='Data Point')
plt.title("Plot with customization")
plt.xlabel("X label")
plt.ylabel("Y label")
plt.xticks(np.arange(0,6,step=1))
plt.yticks(np.arange(0,26,step=5))
plt.legend()
plt.show()
```
