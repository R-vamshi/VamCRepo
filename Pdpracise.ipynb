```python
data = {'region':['East','West','East','West','East','West'],'Product':['A','A','B','B','C','C'],'Sales':[100,150,200,250,300,350]}
df2 = pd.DataFrame(data)
```

```python
df2.pivot_table(values='Sales',index='region',columns='Product',aggfunc='sum')
```

```python
def cust_agg(x):
    return x.max()-x.min()

pivot = df2.pivot_table(values='Sales',index='region',columns='Product',aggfunc=cust_agg)
print(pivot)
```

```python
cross_table = pd.crosstab(df2['region'],df2['Product'])
print(cross_table)
```

```python
data1 = {
'Region': ['East', 'West'],
'Product_A': [100, 150],
'Product_B': [200, 250],
'Product_C': [300, 350]
}
df3 = pd.DataFrame(data1)
# Melting the DataFrame
melted_df = pd.melt(df3, id_vars=['Region'], value_vars=['Product_A', 'Product_B',
'Product_C'], var_name='Product', value_name='Sales')
print(melted_df)
```
